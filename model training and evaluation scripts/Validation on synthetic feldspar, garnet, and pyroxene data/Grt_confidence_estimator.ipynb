# Defining grt end-members

```python
# chemical format SiO2, TiO2, Al2O3, FeO, MnO, MgO, CaO, Na2O, K2O, P2O5, CO2
alm =     np.array([3,   0,    1,     3,   0,   0,   0,    0,   0,   0,   0])
py =      np.array([3,   0,    1,     0,   0,   3,   0,    0,   0,   0,   0])
sps =      np.array([3,   0,   1,     0,   3,  0,    0,    0,   0,   0,   0])

alm = 100 * alm / sum(alm)
py = 100 * py / sum(py)
sps = 100 * sps / sum(sps)
```

```python
def make_almpy(interval = 0.1):
    alm =     np.array([3,   0,    1,     3,   0,   0,   0,    0,   0,   0,   0])
    py =      np.array([3,   0,    1,     0,   0,   3,   0,    0,   0,   0,   0])
    alm = 100 * alm / sum(alm)
    py = 100 * py / sum(py)
    femg = np.array([])
    for i in np.arange(0,1 + interval,interval):
        s = round(i,2)
        comp = (alm * s) + (py * (1 - s))
        # print(comp)
        if (i == 0):
            femg = comp
        else:
            femg = np.vstack((femg,comp))
    femg = np.round(femg,4)
    return(femg)

def make_grt(interval = 0.1):
    femg = make_almpy(interval)
    sps =      np.array([3,   0,   1,     0,   3,  0,    0,    0,   0,   0,   0])
    sps = 100 * sps / sum(sps)
    c = 0
    grt = np.array([])
    for i in np.arange(0,np.shape(femg)[0]):
        for j in np.arange(0,1 + interval, interval):
            comp = sps * j + (femg[i] * (1 - j))
            c = c + 1
            if (c == 1):
                grt = comp
            else:
                grt = np.vstack((grt,comp))
    grt = np.round(grt,4)
    return(grt)
```
